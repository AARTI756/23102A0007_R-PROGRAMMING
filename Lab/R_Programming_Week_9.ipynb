## 2. `cat()` function

`cat()` concatenates its arguments and prints them as continuous output. By default, it separates items with a space. A newline can be supplied using `\n`.

```r
cat("The zero occurs at", 2*pi, "radians.", "\n")
```

```r
# Date example from the notes
d = date()
cat("Today's date is:", d, "\n")
```

```r
# Using sep with a vector
x = 1:10
x

cat(x, sep = " ++ ")
cat("\n")
cat(x, sep = " / ")
cat("\n")
```

```r
# Printing a calculated value
x = 7
cat("The square of", x, "is", x^2, "!\n")
```

```r
# Using format() inside cat()
cat(
  "The square root of", x,
  "is approximately",
  format(sqrt(x), digits = 3),
  "\n"
)
```

```r
# cat() with a simple vector
evenno = c(2, 4, 6, 8, 10)

cat("The first few even numbers are:", evenno, "...\n")
```

```r
# cat() with fill and labels
x = 1:10

cat(
  x,
  fill = 2,
  labels = paste("(", letters[1:10], "):")
)
```

## 3. `paste()` function

`paste()` concatenates strings and returns the result, so the result can be assigned to a variable.  
`sep` controls the separator between input arguments, while `collapse` combines the resulting elements into one string.

```r
# Basic paste()
paste("Everybody", "loves", "R Programming.")
```

```r
# Different separators
paste("Everybody", "loves", "R Programming.", sep = "*")

paste("Everybody", "loves", "R Programming.", sep = "===")
```

```r
# paste() converts its arguments to character strings
paste(1:12)

# Alternative shown in the notes
as.character(1:12)
```

```r
# Vector of names
names = c("Prof. Singh", "Mr. Venkat", "Dr. Jha")

names

paste(names, "is", "a good", "person.")
```

```r
# Using collapse to join the generated strings
paste(
  names,
  "is",
  "a good",
  "person.",
  collapse = ", and "
)
```

## 4. Operations with strings using `paste()`

The notes demonstrate how spaces inside quoted strings are preserved and how `sep` affects the output.

```r
paste(
  1, " is first",
  2, " is second",
  3, " is third",
  sep = "#"
)
```

```r
# Notice the spaces inside the quoted strings
paste(
  1, " is first ",
  2, " is second ",
  3, " is third ",
  sep = "#"
)
```

```r
# Creating a vector of strings
x = paste("Ex", 1:5, sep = "_")

x

x[1]
x[2]
x[3]
x[5]
```

```r
# Using collapse creates one single string
x = paste("Ex", 1:5, sep = "_", collapse = "")

x
x[1]
```

### Difference between `sep` and `collapse`

Without `collapse`, `paste()` produces a vector of strings. With `collapse`, those resulting strings are joined into one string.

```r
# Vector result
x = paste("Ex", 1:5, sep = "_")
x
```

```r
# Single collapsed string
x = paste("Ex", 1:5, sep = "_", collapse = "")
x
```

```r
# paste0() and paste() with a single vector
paste0(1:10)
paste(1:10)
```

```r
# Vectorized paste0()
paste0(
  1:10,
  c("st", "nd", "rd", rep("th", 7))
)
```

```r
# paste() inserts a space by default
paste(
  1:10,
  c("st", "nd", "rd", rep("th", 7))
)
```

## 5. `strsplit()` – String Splitting

`strsplit()` splits each element of a character vector using the supplied `split` pattern. The result is a list.

```r
# Example from the NPTEL notes
x = "The&!syntax&!of&!paste&!is!&available! &inthe online-help"

x

strsplit(x, split = "!")
```

```r
# Splitting using "&!"
x = "The&!syntax&!of&!paste&!is!&available! &inthe online-help"

x

strsplit(x, split = "&!")
```

```r
# Splitting using "!&"
x = "The&!syntax&!of&!paste&!is!&available! &inthe online-help"

x

y = strsplit(x, split = "!&")

y
```

```r
# Accessing individual components of the list returned by strsplit()
y[[1]][1]
y[[1]][2]
y[[1]][3]
```

## 6. `strsplit()` – Convert dates into a matrix

The notes demonstrate splitting date strings using `-`, then converting the resulting list into a matrix.

```r
dates = c(
  "2020-07-24",
  "2021-08-25",
  "2022-09-26",
  "2023-10-27"
)

dates
```

```r
# Split each date into year, month and day
datesplt = strsplit(dates, "-")

datesplt
```

```r
# Matrix with character values
datemat = matrix(
  unlist(datesplt),
  nrow = 4,
  ncol = 3,
  byrow = TRUE
)

datemat
```

```r
# Matrix with numeric values
datematrix = matrix(
  as.numeric(unlist(datesplt)),
  nrow = 4,
  ncol = 3,
  byrow = TRUE
)

datematrix
```

```r
# Splitting a word into individual characters
strsplit("Shalabh", split = "")
```

## 7. Counting characters with `nchar()`

`nchar()` returns the number of characters in each element of its input. `nzchar()` checks whether elements are non-empty strings.

```r
x = "R course 24.07.2022"
y = "Number of participants: 25"

nchar(x)
nchar(y)
```

```r
# nzchar() returns TRUE for non-empty strings
x = "R course 24.07.2022"
y = "Number of participants: 25"

nzchar(x)
nzchar(y)
```

```r
# nzchar() with a character vector
x = c("Apple", "Banana", "Cake")

x
nzchar(x)
```

```r
# Empty string example
y = c("Apple", "", "Cake")

y
nzchar(y)
```

```r
# nchar() with character vectors
x = c("Apple", "Banana", "Cake")

nchar(x)
```

```r
# nchar() can also operate on numeric values
y = c(2, 4, 6)
nchar(y)

z = c(11, 222, 3333)
nchar(z)

z1 = c(1.1, 2.22, 3.333)
nchar(z1)
```

## 8. Changing lower and upper case

`tolower()` converts uppercase letters to lowercase, while `toupper()` converts lowercase letters to uppercase. Non-alphabetic characters remain unchanged.

```r
x = "R course will start from 24.07.2022"

toupper(x)
```

```r
z = "INDIAN INSTITUTE OF TECHNOLOGY"

tolower(z)
```
